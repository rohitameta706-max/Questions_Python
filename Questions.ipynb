```python
# Simple Calculator
a = float(input("Write the first number"))
opt = input("Write the operation to be performed")
b = float(input("Write the second number"))
if opt == '+':
   print("addition of two number is" , a + b)
elif opt == '-'
    print("sub of two numbers is" , a - b)
elif opt == '*'
    print("Multiply of two numbers is" , a * b) 
elif opt == '/'
    print("Division of two numbers is" , a / b) 
else
    print("Cannot do operation")
```
